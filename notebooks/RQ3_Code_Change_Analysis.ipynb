```python
# Setup for Code Change Analysis
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import requests
import re
import os
import sys
from datetime import datetime

# Add src directory to path
sys.path.append('../src')
from data_loader import load_aidev

print("🧰 RQ3: Code Change Analysis")
print("=" * 50)
print(f"📅 Analysis Date: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

# Load sample data
print("\n📂 Loading sample data...")
df = load_aidev(sample_size=100)  # Start small for API testing
print(f"✅ Loaded {len(df):,} PRs for GitHub API analysis")

# GitHub API Setup (placeholder - requires authentication)
def setup_github_api():
    """Setup GitHub API client (requires token)"""
    # Note: In production, use environment variable for token
    # github_token = os.getenv('GITHUB_TOKEN')
    github_token = None  # Placeholder
    
    if not github_token:
        print("⚠️ GitHub token not configured")
        print("📋 To enable GitHub API:")
        print("   1. Create personal access token at https://github.com/settings/tokens")
        print("   2. Set GITHUB_TOKEN environment variable")
        print("   3. Re-run this analysis")
        return None
    
    headers = {
        'Authorization': f'token {github_token}',
        'Accept': 'application/vnd.github.v3+json'
    }
    return headers

# Extract GitHub info from URLs
def extract_github_info(html_url):
    """Extract owner/repo/pr_number from GitHub URL"""
    if pd.isna(html_url):
        return None, None, None
    
    # Pattern: https://github.com/owner/repo/pull/number
    match = re.match(r'https://github\.com/([^/]+)/([^/]+)/pull/(\d+)', str(html_url))
    if match:
        return match.group(1), match.group(2), match.group(3)
    return None, None, None

# Test URL extraction
print("\n🔍 Testing GitHub URL extraction...")
sample_urls = df['html_url'].dropna().head(3)
for url in sample_urls:
    owner, repo, pr_num = extract_github_info(url)
    print(f"  URL: {url}")
    print(f"  Parsed: {owner}/{repo}#{pr_num}")

headers = setup_github_api()
api_available = headers is not None
print(f"\n🔗 GitHub API Status: {'✅ Ready' if api_available else '❌ Not configured'}")
```
