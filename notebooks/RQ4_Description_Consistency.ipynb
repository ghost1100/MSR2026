```python
%pip install scikit-learn

# Setup for Description Consistency Analysis
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import os
import sys
from datetime import datetime
from collections import Counter
import warnings
warnings.filterwarnings('ignore')

# Add src directory to path
sys.path.append('../src')
from data_loader import load_aidev

print("🧠 RQ4: Description Consistency Analysis")
print("=" * 50)
print(f"📅 Analysis Date: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

# Try to import NLP libraries (install if needed)
try:
    import nltk
    from sklearn.feature_extraction.text import TfidfVectorizer
    from sklearn.metrics.pairwise import cosine_similarity
    print("✅ NLP libraries available")
    nlp_available = True
except ImportError:
    print("⚠️ NLP libraries not installed")
    print("📋 To enable full NLP analysis:")
    print("   pip install nltk scikit-learn")
    print("   python -c \"import nltk; nltk.download('punkt'); nltk.download('stopwords')\"")
    nlp_available = False

# Load sample data
print("\n📂 Loading sample data...")
df = load_aidev(sample_size=2000)  # Medium sample for text analysis
print(f"✅ Loaded {len(df):,} PRs for description analysis")

# Basic text statistics
def analyze_text_stats(df):
    """Calculate basic text statistics"""
    stats = {}
    
    # Title statistics
    df['title_length'] = df['title'].str.len()
    df['title_words'] = df['title'].str.split().str.len()
    
    # Body statistics  
    df['body_length'] = df['body'].str.len()
    df['body_words'] = df['body'].str.split().str.len()
    
    # Calculate ratios
    df['title_body_ratio'] = df['title_length'] / (df['body_length'] + 1)  # +1 to avoid div by zero
    
    stats['title'] = {
        'avg_length': df['title_length'].mean(),
        'avg_words': df['title_words'].mean(),
        'median_length': df['title_length'].median()
    }
    
    stats['body'] = {
        'avg_length': df['body_length'].mean(),
        'avg_words': df['body_words'].mean(),
        'median_length': df['body_length'].median()
    }
    
    return df, stats

df_analyzed, text_stats = analyze_text_stats(df)

print(f"\n📊 TEXT STATISTICS:")
print(f"  Title - Avg Length: {text_stats['title']['avg_length']:.1f} chars, Avg Words: {text_stats['title']['avg_words']:.1f}")
print(f"  Body  - Avg Length: {text_stats['body']['avg_length']:.1f} chars, Avg Words: {text_stats['body']['avg_words']:.1f}")

print(f"\n🔍 NLP Analysis Status: {'✅ Ready' if nlp_available else '❌ Limited (basic stats only)'}")
```

```python
# Enhanced setup with Claude API integration
from claude_analyzer import ClaudeAnalyzer, create_claude_enhanced_analysis

# Initialize Claude analyzer for intelligent consistency analysis
try:
    claude_analyzer = ClaudeAnalyzer()
    CLAUDE_AVAILABLE = True
    print("🤖 Claude API integration ready for enhanced consistency analysis!")
    print("💡 This will provide AI-powered description quality assessment")
except Exception as e:
    CLAUDE_AVAILABLE = False
    print(f"⚠️  Claude API not available: {e}")
    print("📊 Falling back to statistical analysis only")
```

```python
# 🤖 CLAUDE-ENHANCED CONSISTENCY ANALYSIS
# This section uses AI to provide intelligent assessment of PR description quality

if CLAUDE_AVAILABLE:
    print("🚀 Starting Claude-Enhanced Description Consistency Analysis...")
    print("💡 This will analyze description quality, consistency, and agent patterns")
    
    # Analyze a sample of PRs for detailed consistency metrics
    sample_size = min(500, len(df))  # Adjust based on needs and API costs
    sample_df = df.sample(n=sample_size, random_state=42)
    
    print(f"📊 Analyzing {sample_size} PRs for consistency patterns...")
    
    # Batch analyze consistency
    consistency_results = []
    
    for idx, (_, row) in enumerate(sample_df.iterrows()):
        if idx % 50 == 0:
            print(f"   Progress: {idx}/{sample_size} ({idx/sample_size:.1%})")
        
        try:
            result = claude_analyzer.analyze_description_consistency(
                title=row['title'],
                description=str(row['body'])
            )
            
            consistency_results.append({
                'agent': row['agent'],
                'consistency_score': result['consistency_score'],
                'reasoning': result['reasoning'],
                'issues_count': len(result['potential_issues']),
                'quality_indicators': len(result['quality_indicators']),
                'potential_issues': result['potential_issues'],
                'title_length': len(row['title']),
                'body_length': len(str(row['body']))
            })
            
        except Exception as e:
            print(f"Warning: Error analyzing PR {idx}: {e}")
            continue
    
    # Convert to DataFrame for analysis
    consistency_df = pd.DataFrame(consistency_results)
    
    print(f"✅ Completed Claude analysis of {len(consistency_df)} PRs")
    
    # Overall consistency insights
    print(f"\n📊 CLAUDE CONSISTENCY INSIGHTS:")
    print(f"   Average Consistency Score: {consistency_df['consistency_score'].mean():.1f}/10")
    print(f"   High Quality PRs (score ≥ 8): {(consistency_df['consistency_score'] >= 8).sum()}/{len(consistency_df)} ({(consistency_df['consistency_score'] >= 8).mean():.1%})")
    print(f"   Low Quality PRs (score ≤ 4): {(consistency_df['consistency_score'] <= 4).sum()}/{len(consistency_df)} ({(consistency_df['consistency_score'] <= 4).mean():.1%})")
    print(f"   Average Issues per PR: {consistency_df['issues_count'].mean():.1f}")
    print(f"   Average Quality Indicators: {consistency_df['quality_indicators'].mean():.1f}")
    
    # Agent-specific consistency analysis
    print(f"\n🤖 AGENT CONSISTENCY PATTERNS:")
    agent_consistency = consistency_df.groupby('agent').agg({
        'consistency_score': ['mean', 'std', 'count'],
        'issues_count': 'mean',
        'quality_indicators': 'mean'
    }).round(2)
    
    for agent in agent_consistency.index:
        score = agent_consistency.loc[agent, ('consistency_score', 'mean')]
        count = agent_consistency.loc[agent, ('consistency_score', 'count')]
        issues = agent_consistency.loc[agent, ('issues_count', 'mean')]
        quality = agent_consistency.loc[agent, ('quality_indicators', 'mean')]
        
        print(f"   {agent}: Score {score:.1f}/10 (n={count}), Issues: {issues:.1f}, Quality: {quality:.1f}")
    
else:
    print("⚠️  Skipping Claude analysis - API not available")
    print("📊 Using statistical analysis only")
```
